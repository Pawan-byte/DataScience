```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
sigmoid(0),sigmoid(1.99),sigmoid(-1.99),sigmoid(100),sigmoid(-100)
```

```python
#Visualise sigmoid function
x = np.linspace(-10,10,100)
y = sigmoid(x)
plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('sigmoid(x)')
plt.grid()
plt.show()
```

Log loss

```python
def log_loss(y_true, y_pred):
    loss = y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred)
    return -np.mean(loss)
```
